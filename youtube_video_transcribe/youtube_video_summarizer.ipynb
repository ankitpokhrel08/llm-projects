```python
def get_video_id(url):
  try:
    return url.split('?v=')[1]
  except:
    return (url.split('/')[3]).split('?')[0]
```

```python
from youtube_transcript_api import YouTubeTranscriptApi

ytt_api = YouTubeTranscriptApi()
transcript = ytt_api.fetch(video_id)
transcript
```

```python
transcript_joined = ' '.join([line.text for line in transcript])
transcript_joined
```

```python
import google.generativeai as genai
from google.colab import userdata

# Configure Gemini API with the key from Colab secrets
GOOGLE_API_KEY = "your_google_api_key_here"  # Replace with your actual API key
genai.configure(api_key=GOOGLE_API_KEY)
```

```python
# Initialize the Gemini Pro model (suitable for text summarization)
gemini_model = genai.GenerativeModel('gemini-2.5-flash')
```

```python
def summarize_text_with_gemini(text):
    prompt = f"Please provide a concise summary of the following text:\n\n{text}"
    response = gemini_model.generate_content(prompt)
    return response.text

# Get the summary of the joined transcript
summary = summarize_text_with_gemini(transcript_joined)
print(summary)
```
